```python
@docker()
@mp()
def compute_kl_divergence(p, q):

    """
    Compute the Kullback-Leibler divergence D(p || q) for discrete distributions p and q.

    Parameters:
    p (np.ndarray): Probability distribution p.
    q (np.ndarray): Probability distribution q.

    Returns:
    float: The KL divergence D(p || q).
    """
    # Ensure p and q are numpy arrays
    p = np.asarray(p)
    q = np.asarray(q)

    # Avoid division by zero and log of zero
    p = np.clip(p, 1e-10, None)
    q = np.clip(q, 1e-10, None)

    return np.sum(p * np.log(p / q))
```
