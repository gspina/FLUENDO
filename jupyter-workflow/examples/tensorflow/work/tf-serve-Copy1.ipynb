```python
# TensorFlow and tf.keras
import tensorflow as tf
from tensorflow import keras


fashion_mnist = keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

# scale the values to 0.0 to 1.0
train_images = train_images / 255.0
test_images = test_images / 255.0

# reshape for feeding into the model
train_images = train_images.reshape(train_images.shape[0], 28, 28, 1)
test_images = test_images.reshape(test_images.shape[0], 28, 28, 1)

class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

print('\ntrain_images.shape: {}, of {}'.format(train_images.shape, train_images.dtype))
print('test_images.shape: {}, of {}'.format(test_images.shape, test_images.dtype))
```

```python
import numpy as np

# Numero totale di classi (Fashion MNIST ha 10 classi: 0–9)
NUM_CLASSES = 10
HALF = NUM_CLASSES // 2   # =5

# Primo sottoinsieme di classi: {0,1,2,3,4}
classes_1 = list(range(0, HALF))
# Secondo sottoinsieme di classi: {5,6,7,8,9}
classes_2 = list(range(HALF, NUM_CLASSES))

# --- Funzione per estrarre esempi la cui label è in `class_list`, rimappando le etichette su 0..(len(class_list)-1) ---
def extract_subset(images, labels, class_list):
    """
    Filtra images, labels tenendo solo le etichette presenti in class_list;
    rimappa le etichette in un range 0..(len(class_list)-1).
    """
    mask = np.isin(labels, class_list)
    imgs_sub = images[mask]
    lbls_sub = labels[mask]
    # Rimappo ogni label x in label_index[x] dove label_index mappa class_list su 0..len-1
    label_index = {c:i for i,c in enumerate(class_list)}
    lbls_sub_mapped = np.vectorize(lambda x: label_index[x])(lbls_sub)
    return imgs_sub, lbls_sub_mapped

# Estrazione per il “client 1” (classi 0–4)
train_images_1, train_labels_1 = extract_subset(train_images, train_labels, classes_1)
test_images_1,  test_labels_1  = extract_subset(test_images,  test_labels,  classes_1)

# Estrazione per il “client 2” (classi 5–9)
train_images_2, train_labels_2 = extract_subset(train_images, train_labels, classes_2)
test_images_2,  test_labels_2  = extract_subset(test_images,  test_labels,  classes_2)

print("Client 1 – esempi di train:", train_images_1.shape, train_labels_1.shape)
print("Client 2 – esempi di train:", train_images_2.shape, train_labels_2.shape)
```

```python
import tensorflow as tf
from tensorflow import keras

def create_local_model(num_local_classes=5):
    """
    Restituisce un modello convolutional semplice con output = num_local_classes.
    """
    m = keras.Sequential([
        keras.layers.Conv2D(input_shape=(28,28,1), filters=8, kernel_size=3, 
                            strides=2, activation='relu', name='Conv1'),
        keras.layers.Flatten(),
        keras.layers.Dense(num_local_classes, activation='softmax', name='Softmax')
    ])
    return m

# Creo i due modelli
model_1 = create_local_model(num_local_classes=HALF)   # output=5, per classi [0-4]
model_2 = create_local_model(num_local_classes=HALF)   # output=5, per classi [5-9]

print("=== Model 1 (classi 0..4) ===")
model_1.summary()
print("\n=== Model 2 (classi 5..9) ===")
model_2.summary()
```

```python
# Numero di epoche
EPOCHS_LOCAL = 10   # puoi aumentare se vuoi

# ---> Model 1 su train_images_1 / train_labels_1 <---
model_1.compile(optimizer='adam',
                loss='sparse_categorical_crossentropy',
                metrics=['accuracy'])
model_1.fit(train_images_1, train_labels_1, epochs=EPOCHS_LOCAL, validation_split=0.1)
loss1, acc1 = model_1.evaluate(test_images_1, test_labels_1)
print(f"\nModel1 (classi 0–4) – Test accuracy: {acc1:.4f}")

# ---> Model 2 su train_images_2 / train_labels_2 <---
model_2.compile(optimizer='adam',
                loss='sparse_categorical_crossentropy',
                metrics=['accuracy'])
model_2.fit(train_images_2, train_labels_2, epochs=EPOCHS_LOCAL, validation_split=0.1)
loss2, acc2 = model_2.evaluate(test_images_2, test_labels_2)
print(f"\nModel2 (classi 5–9) – Test accuracy: {acc2:.4f}")
```

```python
# --- Definisco il modello globale (tutta la rete, con output=10) ---
def create_global_model(num_classes=NUM_CLASSES):
    m = keras.Sequential([
        keras.layers.Conv2D(input_shape=(28,28,1), filters=8, kernel_size=3,
                            strides=2, activation='relu', name='Conv1'),
        keras.layers.Flatten(),
        keras.layers.Dense(num_classes, activation='softmax', name='Softmax')
    ])
    return m

global_model = create_global_model()

# --- Estrazione dei pesi dai modelli locali ---
# Modello 1: conv + dense
conv_w1, conv_b1 = model_1.layers[0].get_weights()
dense_w1, dense_b1 = model_1.layers[2].get_weights()  # layer index 2 = Dense

# Modello 2: conv + dense
conv_w2, conv_b2 = model_2.layers[0].get_weights()
dense_w2, dense_b2 = model_2.layers[2].get_weights()

# --- Calcolo dei nuovi pesi per global_model ---
# 1) Strato conv: media aritmetica
conv_w_global = (conv_w1 + conv_w2) / 2.0
conv_b_global = (conv_b1 + conv_b2) / 2.0

# 2) Strato Dense: concateno orizzontalmente
#    dense_w1 e dense_w2 hanno shape (neuroni_flatten, 5)
dense_w_global = np.concatenate([dense_w1, dense_w2], axis=1)   # shape = (neuroni_flatten, 10)
dense_b_global = np.concatenate([dense_b1, dense_b2], axis=0)   # shape = (10,)

# Controllo dimensioni
print("Shape conv_w_global:", conv_w_global.shape)
print("Shape conv_b_global:", conv_b_global.shape)
print("Shape dense_w_global:", dense_w_global.shape)
print("Shape dense_b_global:", dense_b_global.shape)

# --- Imposto i pesi sul global_model ---
global_model.layers[0].set_weights([conv_w_global, conv_b_global])
global_model.layers[2].set_weights([dense_w_global, dense_b_global])

print("\n=== Global model (inizializzato) ===")
global_model.summary()
```

```python
import tensorflow_federated as tff
import tensorflow as tf

# 0) Assicurati che create_global_model() sia già definita in una cella precedente:
#    def create_global_model(num_classes):
#        return tf.keras.Sequential([
#            tf.keras.layers.Conv2D(input_shape=(28,28,1), filters=8, kernel_size=3,
#                                   strides=2, activation='relu', name='Conv1'),
#            tf.keras.layers.Flatten(),
#            tf.keras.layers.Dense(num_classes, activation='softmax', name='Softmax')
#        ])

# 1) Parametri federated
NUM_ROUNDS = 5
BATCH_SIZE = 32

# 2) Funzione di preprocess (batch + shuffle)
def preprocess(dataset):
    return (
        dataset
        .shuffle(buffer_size=1000)
        .batch(BATCH_SIZE)
        .prefetch(tf.data.AUTOTUNE)
    )

# 3) Creo i dataset federati per i due client
ds_client_1 = tf.data.Dataset.from_tensor_slices((train_images_1, train_labels_1))
ds_client_1 = preprocess(ds_client_1)

ds_client_2 = tf.data.Dataset.from_tensor_slices((train_images_2, train_labels_2))
ds_client_2 = preprocess(ds_client_2)

federated_train_data = [ds_client_1, ds_client_2]

# 4) Prelevo example_spec da ds_client_1.element_spec (serve come input_spec in TFF)
example_spec = ds_client_1.element_spec

# 5) Definisco create_keras_model_for_tff(): deve restituire un Keras identico a global_model
def create_keras_model_for_tff():
    return create_global_model(num_classes=NUM_CLASSES)

# 6) Costruisco model_fn() che crea il wrapper TFF a partire da un Keras “vergine”
def model_fn():
    keras_model = create_keras_model_for_tff()
    return tff.learning.models.from_keras_model(
        keras_model=keras_model,
        input_spec=example_spec,
        loss=tf.keras.losses.SparseCategoricalCrossentropy(),
        metrics=[tf.keras.metrics.SparseCategoricalAccuracy()]
    )

# 7) Build del processo di Federated Averaging (weighted), passando OGGETTI SGD di TFF
iterative_process = tff.learning.algorithms.build_weighted_fed_avg(
    model_fn=model_fn,
    client_optimizer_fn=lambda: tff.learning.optimizers.SGD(learning_rate=0.02),
    server_optimizer_fn=lambda: tff.learning.optimizers.SGD(learning_rate=1.0)
)

# 8) Inizializzo lo stato federato
state = iterative_process.initialize()

# 9) Sovrascrivo i pesi iniziali (random) con quelli di global_model
model_weights_from_keras = tff.learning.ModelWeights.from_model(global_model)
state = state._replace(model_weights=model_weights_from_keras)

# 10) Ciclo di training federato
for round_num in range(1, NUM_ROUNDS + 1):
    result = iterative_process.next(state, federated_train_data)
    state = result.state
    print(f"Round {round_num} – metriche aggregate: {result.metrics}")

# 11) Creo un nuovo Keras model e ci carico i pesi finali
final_model = create_global_model(num_classes=NUM_CLASSES)
final_model.set_weights([w.numpy() for w in state.model_weights.trainable])

# 12) Valutazione su tutto il test set (10 classi originali)
final_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
loss_all, acc_all = final_model.evaluate(test_images, test_labels)
print(f"\nGlobal federated model – test accuracy (10 classi): {acc_all:.4f}")
```
